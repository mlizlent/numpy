```python
import numpy as np

# Generate a mock 10x10 score difference forecast matrix
generator = np.random.default_rng(42)
mock_matrix = generator.uniform(-4.0, 4.0, size=(10, 10))

# A team vs itself has a score difference of 0
np.fill_diagonal(mock_matrix, 0.0)

# Save it to your current working directory
np.savetxt('model_forecasts.txt', mock_matrix)
print("Successfully generated a mock 'model_forecasts.txt' file!")
```

```python
import itertools
import numpy as np

#Load the forecast matrix
model_forecasts = np.loadtxt('model_forecasts.txt')

#  Generate all permutations for the 10 teams (0 to 9)
all_perms = np.array(list(itertools.permutations(range(10))))

#Split permutations into Team 1 and Team 2 pairs for the 5 matches
t1_indices = all_perms[:, 0::2]  
t2_indices = all_perms[:, 1::2] 
# Use advanced matrix indexing to pull forecasts for all pairs simultaneously
match_forecasts = model_forecasts[t1_indices, t2_indices]

# Calculate the sum of squared differences for each pairing scenario
sum_squared_diffs = np.sum(match_forecasts**2, axis=1)

# Locate the scenario that minimizes this sum
best_scenario_idx = np.argmin(sum_squared_diffs)
best_permutation = all_perms[best_scenario_idx]

# Format the output into the expected 2D matrix structure
optimal_pairing = np.array([best_permutation[0::2], best_permutation[1::2]])

print("Optimal Pairings Matrix:")
print(optimal_pairing)
```
