```python
import numpy as np
matrix = np.ones((9, 9), dtype=np.int8)
print("---Initial 9x9 Matrix---")
print(matrix)
print()


matrix[1:8, 1:8] = 0
matrix[2:7, 2:7] = 1
matrix[3:6, 3:6] = 0
matrix[4, 4] = 1

print("---concentric Parttern matrix---")
print(matrix)
print()

arr1 = np.array([1, 2, 3, 4, 5], dtype=np.int8)
arr2 = np.array([1, 2, 3], dtype=np.int8)

broadcasting_output = arr1[:, np.newaxis] *arr2

print("--- Broadcasting Multiplication Matrix ---")
print(broadcasting_output)
```
