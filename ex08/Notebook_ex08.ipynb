```python
import numpy as np

#Load the data using genfromtxt and optimize size to float32
data = np.genfromtxt('ex08/data/winequality-red.csv', delimiter=';', skip_header=1, dtype=np.float32)

# Verify the byte size of the array
print(f"Array size in bytes: {data.nbytes}") 


#Display the 2nd, 7th, and 12th rows as a two-dimensional array
selected_rows = data[[1, 6, 11], :]
# Exclude NaN values if any are present
selected_rows = selected_rows[~np.isnan(selected_rows).any(axis=1)]
print("\nSelected rows (2nd, 7th, 12th):\n", selected_rows)


# Determine if any wine has an alcohol percentage > 20%
alcohol_col = data[:, 10]
any_above_20 = np.any(alcohol_col > 20.0)
print(f"\nAny wine with alcohol > 20%: {any_above_20}")


# Calculate the average alcohol percentage excluding NaNs
avg_alcohol = np.nanmean(alcohol_col)
print(f"Average alcohol percentage: {avg_alcohol:.2f}%")


# Compute statistical measures for pH values
ph_col = data[:, 8]
ph_min = np.nanmin(ph_col)
ph_max = np.nanmax(ph_col)
ph_25 = np.nanpercentile(ph_col, 25)
ph_50 = np.nanpercentile(ph_col, 50)
ph_75 = np.nanpercentile(ph_col, 75)
ph_mean = np.nanmean(ph_col)

print("\n--- pH Statistical Measures ---")
print(f"Minimum: {ph_min}")
print(f"Maximum: {ph_max}")
print(f"25th Percentile: {ph_25}")
print(f"50th Percentile (Median): {ph_50}")
print(f"75th Percentile: {ph_75}")
print(f"Mean: {ph_mean:.4f}")


#Find average quality score of wines with the 20% least sulphate content
sulphates_col = data[:, 9]
quality_col = data[:, 11]

sulphate_20_percentile = np.nanpercentile(sulphates_col, 20)
low_sulphate_mask = sulphates_col < sulphate_20_percentile
avg_quality_low_sulphate = np.nanmean(quality_col[low_sulphate_mask])
print(f"\nAvg quality score (lowest 20% sulphates): {avg_quality_low_sulphate:.4f}")


# Compute the mean of all variables for best and worst quality wines
max_quality = np.nanmax(quality_col)
min_quality = np.nanmin(quality_col)

best_quality_mask = (quality_col == max_quality)
worst_quality_mask = (quality_col == min_quality)

mean_best_quality = np.nanmean(data[best_quality_mask], axis=0)
mean_worst_quality = np.nanmean(data[worst_quality_mask], axis=0)

print("\nMean of variables for BEST quality wines:\n", mean_best_quality)
print("\nMean of variables for WORST quality wines:\n", mean_worst_quality)
```
